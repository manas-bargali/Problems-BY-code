```python
def prime_or_not(number):
    # traditional approach 
    if number < 2:
        return f"{number} is not a prime number"
    elif number == 2:
        return f"{number} is even prime number only "
    else:
        for i in range(2,number):
            if number%i==0:
                return f"{number} is not a prime number"
        return f"{number} is odd prime number."
```

```python
prime_or_not(157)
```
